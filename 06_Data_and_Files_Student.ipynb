## Plain text file

We have seen this before.
- A sequence of characters with no structure
>- or you can create your own "structure"
- Each line is separated by \n
- There are no types — everything is a string

**When to use:**

logs, simple config, freeform notes, any data without structure.

However, we can add structure to a plain text file ourselves.


For example, `note.txt`

```
Meeting at 9am
Remember to call Alice
Buy groceries
```

Several ways to read data from a plain text file.

````python
# Read entire file as one string
with open('notes.txt', 'r', encoding='utf-8') as f:
    content = f.read()
````

````python
# Read line by line (better for large files)
with open('notes.txt', 'r', encoding='utf-8') as f:
    for line in f:
        print(line.strip())   # .strip() removes the \n at the end
````

````python
# Read all lines into a list
with open('notes.txt', 'r', encoding='utf-8') as f:
    lines = f.readlines()     # ['Meeting at 9am\n', 'Remember...\n', ...]
````

You can create a plain text file to overwrite (w) or to append (a)

````python
# Write (creates file, overwrites if exists)
with open('notes.txt', 'w', encoding='utf-8') as f:
    f.write('Line one\n')
    f.write('Line two\n')
````

````python
# Append (adds to end without deleting existing content)
with open('notes.txt', 'a', encoding='utf-8') as f:
    f.write('New line added\n')
````

## CSV file

- A CSV file represents a **table**
- Each line is one row
- Values in each row are **separated by commas**
- Each value is supposed to be in a column
- The first line is usually the header (column names)

For example, `students.csv`

In the file, data is actually stored like a plain text with commas.
```
name,score,grade
Alice,92,A
Bob,78,B
Charlie,65,C
```

If you open a CSV file in a spreadsheet program, like MS Excel or Google Sheet, you will see the data as a table.

```
Name        Score       Grade
Alice       92          A
Bob         78          B
Charlie     65          C
```

### csv module

**Reading data**

- First, `import csv`
- Use `csv.DictReader()` to read the file row by row
- Data will be in a dictionary with the column name as the key

>For example,
>```
>row["score"]
>```

>This will get a string value of the current row in the column named `score`. You will need to cast the value into the working type.

````python
import csv

# Reading — csv.DictReader gives each row as a dict
with open('students.csv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row)
        # {'name': 'Alice', 'score': '92', 'grade': 'A'}
        # Note: ALL values are strings — convert manually if needed
        score = int(row['score'])
````

**Writing data**

- You might want to put data in a list of dictionary items
>- Each dict contains values of a row
>- keys = column names
- Use the `csv.DictWriter()` to write data into a CSV file.
- must write the header with column names first.

````python
import csv

# Writing — csv.DictWriter
rows = [
    {'name': 'Alice', 'score': 92, 'grade': 'A'},
    {'name': 'Bob',   'score': 78, 'grade': 'B'},
]
with open('students.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=['name', 'score', 'grade'])
    writer.writeheader()
    writer.writerows(rows)
````

### pandas module

We can use the pandas module to work with more complicated data and their operations.

**What is pandas?**

- A Python library for working with tabular data
- Built on top of NumPy, optimized for data analysis
- Install with `pip install pandas`
- Import with `import pandas as pd`

There are different data structures to work with

- Series
- Dataframe

#### Series

- A single column of data with an index
- Like a Python list, but with labels
- The first column is for labels
- The second column is for values

````python
import pandas as pd

s = pd.Series([92, 78, 65], index=['Alice', 'Bob', 'Charlie'])
print(s)

# Alice      92
# Bob        78
# Charlie    65
````

####DataFrame

- A table — rows and columns
- Think of it as a dict of Series, where each key is a column name
- Dataframe is what you get when you call `pd.read_csv()`

**Properties**
- A **single column** accessed from a `DataFrame` returns a `Series`
- Every DataFrame has a row index — by default it's 0, 1, 2, ...
- You can set a column as the row index (e.g. student ID)
>- And use a meaningful data in the column as the row label instead of numbers
- Each column has one data type (int, float, str, bool)
- pandas infers types automatically when reading a CSV
>- unlike the csv module where everything is a string

````python
import pandas as pd

df = pd.DataFrame({
    'name':  ['Alice', 'Bob', 'Charlie'],
    'score': [92, 78, 65],
    'grade': ['A', 'B', 'C']
})

print(df)

#       name  score grade
# 0    Alice     92     A
# 1      Bob     78     B
# 2  Charlie     65     C
````

Now, let's get a CSV file into pandas using `pd.read_csv()`

And we can do diffrent data manipulations with dataframe.

For example, data exploration, statistical analysis, and data filtering.

To write a dataframe back to CSV, use `df.to_csv()`

````python
import pandas as pd

# Reading — returns a DataFrame
df = pd.read_csv('students.csv')

print(df)
#       name  score grade
# 0    Alice     92     A
# 1      Bob     78     B
# 2  Charlie     65     C

print(df.shape)           # (3, 3) — rows, columns
print(df.columns.tolist()) # ['name', 'score', 'grade']
print(df['score'].mean())  # 78.33 — pandas infers numeric types automatically

# Access a column
print(df['name'])          # Series: Alice, Bob, Charlie

# Access a single value
print(df.loc[0, 'name'])   # 'Alice'

# Filter rows
passing = df[df['score'] >= 70]

# Writing back to CSV
df.to_csv('output.csv', index=False)  # index=False removes the row numbers
````

## JSON file

- JSON stores structured, nested data
>- Same shape as Python dicts and lists
- Values have types
>- string
>- number
>- boolean
>- null
>- list
>- nested object
- Just `import json` before use

For example, in the file `student.json`:

```
{
    "name": "Alice",
    "score": 92,
    "passed": true,
    "courses": ["Math", "Physics", "CS"],
    "address": {
        "city": "Bangkok",
        "zip": "10100"
    }
}
```
When loaded into a program, this becomes exactly a Python dict:
```
{
    'name': 'Alice',
    'score': 92,           # int, not string
    'passed': True,        # bool
    'courses': ['Math', 'Physics', 'CS'],
    'address': {'city': 'Bangkok', 'zip': '10100'}
}
```



**When to use**
- app settings/config
- nested or hierarchical data
- API responses
- anything that needs to preserve data types (int, bool, list)

**Reading a JSON file**
- Use `json.load()` to read data in a JSON file into a variable.
- Then you can just use the data like a dictionary!

````python
import json

with open('student.json', 'r', encoding='utf-8') as f:
    data = json.load(f)    # parses JSON → Python dict/list

print(data['name'])           # 'Alice'
print(data['score'])          # 92  (int, not string)
print(data['courses'][0])     # 'Math'
print(data['address']['city']) # 'Bangkok'
````

**Writing a JSON file**

- Use `json.dump()` to write a dictionary into a JSON file

````python
import json

student = {
    'name': 'Alice',
    'score': 92,
    'passed': True,
    'courses': ['Math', 'Physics', 'CS']
}

with open('student.json', 'w', encoding='utf-8') as f:
    json.dump(student, f, indent=4)    # indent=4 makes it human-readable
````

**Practical Use**

- A common pattern for saving app data is to use a list of records.
- Each element of the list is a dictionary containing data for a "row" or a record
- You can just read and write the list of records out and into a JSON file as normally.

````python
import json

# List of dicts — like a CSV but with types preserved
students = [
    {'name': 'Alice', 'score': 92, 'passed': True},
    {'name': 'Bob',   'score': 78, 'passed': True},
]

with open('students.json', 'w', encoding='utf-8') as f:
    json.dump(students, f, indent=4)

# Loading back
with open('students.json', 'r', encoding='utf-8') as f:
    students = json.load(f)   # returns a list of dicts
````

###Handling file paths with pathlib.Path

- `pathlib.Path` is a modern, object-oriented way to work with file paths in Python
- Instead of string manipulation, you get clean, readable path
- Just `from pathlib import Path` and you are good to go!

**Getting a current directory**

- Working directory: `current = Path.cwd()`
>- e.g. /home/user/myproject
- Running directory: `current = Path(__file__).parent`
>- the current folder of .py file that is running, not the one in the terminal
>- `__file__`: the running .py file - /home/user/project/myscript.py
>- `Path(__file__)`: Path object of the file itself - /home/user/project/myscript.py
>- `Path(__file__).parent`: the folder that .py is in - /home/user/project


````python
from pathlib import Path

base = Path(__file__).parent # get the folder
config_path = base / "config.json"  # /home/user/project/config.json
````

###Review: basic file error handling

When working with files, things can go wrong
- the file might not exist
- you might not have permission to read it
-etc

Remember? Use try/except to handle this gracefully



````python
from pathlib import Path

p = Path("data/config.json")

try:
    content = p.read_text()
    print(content)
except FileNotFoundError:
    print(f"File not found: {p}")
except PermissionError:
    print("You don't have permission to read this file.")
except Exception as e:
    print(f"Something went wrong: {e}")
````

The `try` block runs your code, and if an error occurs, Python jumps to the matching `except` block and print the corresponding error message, instead of crashing.
